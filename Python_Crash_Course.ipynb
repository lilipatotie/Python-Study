### 1. **Adding Whitespace to Strings with Tabs or Newlines**
Refers to any nonprinting numbers like spaces, tabs, and end-of-line symbols. People use this to organize the output so that users can read it better.

To organize text and strings better:
*   **for tabs**: use **\t** before text in quotations
*   **for newline**: use **\n** after a text inside quotations

```python
print("\tPython") # adds tab to text
```

```python
print("Languages:\nPython\nC\nJavascript") # adds newline to the string
```

### 2. **Stripping Whitespace**
Extra whitespace can be confusing in programs. Python detects 'python ', it is considered significant unless stated otherwise.

To ensure no whitespace exists:
*   use **.rstrip()** method -> right side
* use **.lstrip()** method -> left side
* use **.strip()** methos -> both sides



```python
favorite_language = 'python '
print(favorite_language) # still has space
print(favorite_language.rstrip()) # space removed
```

### 3. **Removing Prefixes**
Here's how to do that:
* use **.removeprefix()** and insert the text you want to remove inside the parenthesis.

```python
nostarch_url = 'https://nostarch.com'
print(nostarch_url.removeprefix('https://'))

simple_url = nostarch_url.removeprefix('https://')
print(simple_url) # assign a differnt value if you want to remove original string
```

### 3. **Avoiding Syntax Errors with Strings**

Proper use of apostrophes are important to avoid syntax errors.

To avoid these issues:
* Use double quotes **""** when there's a **'** inside the text.
* Use single quotes **''** when there's no single quote **'** inside the text.


```python
message = 'One of Python's strengths is its diverse community.'
print(message)
```

```python
message = "One of Python's strengths is its diverse community."
print(message)
```
